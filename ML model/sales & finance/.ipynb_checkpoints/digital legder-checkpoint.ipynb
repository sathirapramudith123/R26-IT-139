```python
import pandas as pd
import numpy as np


df = pd.read_csv("financial_credit_readiness.csv")
print("Shape:", df.shape)
df.head()
```

```python
print("Missing values per column:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print("Rows after de-duplication:", df.shape[0])
```

```python
df_clean = df.drop(columns=["shop_id", "year_month"])
print("Shape after dropping IDs:", df_clean.shape)   # expect (7020, 11)
df_clean.head()
```

```python
print(df_clean.describe().round(2))
```

```python
X = df_clean.drop(columns=["target_credit_ready"])
y = df_clean["target_credit_ready"]
print("Features:", X.shape, "| Target:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,       # 25% for testing
    random_state=42,      # fixed seed -> reproducible
    stratify=y            # keep the same credit-ready ratio in both sets
)

print("Training set:", X_train.shape, "|", len(X_train), "rows (75%)")
print("Testing set :", X_test.shape, "|", len(X_test), "rows (25%)")
print("\nTrain target balance:\n", y_train.value_counts(normalize=True).round(3))
print("Test target balance:\n",  y_test.value_counts(normalize=True).round(3))
```

```python
from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators=300,      # number of trees
    max_depth=4,           # tree depth (shallow -> avoids overfitting)
    learning_rate=0.08,    # contribution of each tree
    subsample=0.9,         # use 90% of rows per tree (robustness)
    eval_metric="logloss",
    random_state=42
)

model.fit(X_train, y_train)
print("Model trained ✓")
```

```python
from sklearn.metrics import accuracy_score

train_pred = model.predict(X_train)
print("Training accuracy:", round(accuracy_score(y_train, train_pred), 3))
```

```python
from sklearn.metrics import (accuracy_score, roc_auc_score,
                             classification_report, confusion_matrix,
                             ConfusionMatrixDisplay)
import matplotlib.pyplot as plt

y_pred  = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print("Accuracy :", round(accuracy_score(y_test, y_pred), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_proba), 3))
print("\nClassification report:\n",
      classification_report(y_test, y_pred, target_names=["not ready", "ready"]))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred, display_labels=["not ready", "ready"], cmap="Blues")
plt.title("Component 1 — Credit-Readiness")
plt.tight_layout()
plt.show()
```

```python
import shap

explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)

# Bar chart: average impact of each feature
shap.summary_plot(shap_values, X_test, plot_type="bar")
```

```python
shap.summary_plot(shap_values, X_test)
```

```python
i = 0   # change this index to inspect different test merchants
print("Predicted P(credit-ready):", round(float(y_proba[i]), 3))

shap.plots._waterfall.waterfall_legacy(
    explainer.expected_value,
    shap_values[i],
    feature_names=list(X_test.columns)
)
```

```python
import numpy as np

# Pick a 'not ready' merchant who is closest to becoming ready
not_ready = np.where(y_pred == 0)[0]
i = not_ready[np.argmax(y_proba[not_ready])]
row = X_test.iloc[[i]].copy()
base = model.predict_proba(row)[0, 1]

# Improve the key levers
cf = row.copy()
cf["profit_margin_pct"]    += 6
cf["stockout_rate"]         = max(0.0, cf["stockout_rate"].iloc[0] - 0.06)
cf["digital_payment_ratio"] = min(0.6, cf["digital_payment_ratio"].iloc[0] + 0.15)
cf["months_active"]         = cf["months_active"].iloc[0] + 6
new = model.predict_proba(cf)[0, 1]

print(f"P(credit-ready): {base:.2f}  ->  {new:.2f}")
print("Recommended actions to become credit-ready:")
print(" - raise profit margin by ~6 percentage points")
print(" - cut stock-out rate by ~6 points")
print(" - grow digital (LankaQR/card) payments by ~15 points")
print(" - keep recording for ~6 more months")
```

```python
import joblib
joblib.dump(model, "component1_credit_readiness_model.pkl")
print("Model saved as component1_credit_readiness_model.pkl")

# To load it later:
# model = joblib.load("component1_credit_readiness_model.pkl")
```
