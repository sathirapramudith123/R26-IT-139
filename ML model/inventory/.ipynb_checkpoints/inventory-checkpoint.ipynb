```python
import pandas as pd
import numpy as np

df = pd.read_csv("demand_forecast_weekly.csv")
print("Shape:", df.shape)
df.head()
```

```python
print("Missing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print("Rows after de-duplication:", df.shape[0])
```

```python
df_clean = pd.get_dummies(df, columns=["item", "category"], drop_first=False)
print("Shape after encoding:", df_clean.shape)   # columns expand a lot
df_clean.head()
```

```python
X = df_clean.drop(columns=["target_units_sold"])
y = df_clean["target_units_sold"]
print("Features:", X.shape, "| Target:", y.shape)
```

```python
# Train on past years, test on the most recent year (2026) — like real forecasting
train_mask = X["iso_year"] < 2026
X_train, X_test = X[train_mask], X[~train_mask]
y_train, y_test = y[train_mask], y[~train_mask]

print("Train rows (2022–2025):", X_train.shape[0])
print("Test rows (2026):", X_test.shape[0])
```

```python
from xgboost import XGBRegressor

model = XGBRegressor(
    n_estimators=400, max_depth=5, learning_rate=0.05,
    subsample=0.9, random_state=42
)
model.fit(X_train, y_train)
print("Model trained ✓")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

pred = model.predict(X_test)
rmse = mean_squared_error(y_test, pred) ** 0.5

print("MAE :", round(mean_absolute_error(y_test, pred), 2), "units/week")
print("RMSE:", round(rmse, 2), "units/week")
print("R²  :", round(r2_score(y_test, pred), 3))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.scatter(y_test, pred, alpha=0.3)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel("Actual units sold"); plt.ylabel("Predicted units sold")
plt.title("Component 2 — Demand forecast (2026 test)")
plt.tight_layout(); plt.show()
```

```python
import shap
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)
shap.summary_plot(shap_values, X_test, plot_type="bar", max_display=12)
```

```python
import joblib
joblib.dump(model, "component2_demand_forecast_model.pkl")
print("Saved as component2_demand_forecast_model.pkl")
```
